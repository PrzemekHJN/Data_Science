```python
from sklearn.feature_extraction.text import TfidfVectorizer

# tfidf
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(texts)

print("Kształt macierzy:", tfidf_matrix.shape)
```

```python
import nltk
from nltk.tokenize import word_tokenize

# tokenizacja 
nltk.download('punkt')
nltk.download('punkt_tab')

sentences = [word_tokenize(doc.lower()) for doc in texts]

print(sentences[0], sentences[1], sentences[49])
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=50,
    window=3,
    min_count=1
)
```

```python
print("Podobieństwo bitcoin-crypto:",
      model.wv.similarity('bitcoin', 'crypto'))

print("\nSłowa podobne do 'market':")
print(model.wv.most_similar('market'))
```

```python
from gensim.corpora.dictionary import Dictionary

dictionary = Dictionary(sentences)
corpus = [dictionary.doc2bow(text) for text in sentences]
```

```python
from gensim.models import LdaModel

lda = LdaModel(
    corpus,
    num_topics=25,
    id2word=dictionary,
    passes=15
)

for idx, topic in lda.print_topics():
    print(f"Temat {idx}: {topic}")
```

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=2)
X_2d = svd.fit_transform(tfidf_matrix)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_2d[:, 0], X_2d[:, 1], alpha=0.7)

for i, txt in enumerate(texts):
    plt.annotate(i, (X_2d[i, 0], X_2d[i, 1]))

plt.title("Dokumenty krypto w przestrzeni 2D (SVD)")
plt.xlabel("Składowa 1")
plt.ylabel("Składowa 2")
plt.show()
```
