```python
# imports
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
# trening data generation
x = np.linspace(1, 3, 1000)

# column format
X = x.reshape(-1, 1)

# function
y = np.sin(x**2)
y = y.reshape(-1, 1)
```

```python
# neural network model definition
model = Sequential([
    Input(shape=(1,)),
    Dense(64, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1)
])
```

```python
# model compilation
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
# network training
history = model.fit(
    X,
    y,
    epochs=200,
    batch_size=32,
    verbose=1
)
```

```python
# value prediction
y_pred = model.predict(X)
```

```python
# visualization
plt.figure(figsize=(7,7))

# scatter plot
plt.scatter(
    y,
    y_pred,
    alpha=0.6,
    s=10,
    edgecolors='black',
    label="Predykcje modelu"
)

# perfect fit line
plt.plot(
    [y.min(), y.max()],
    [y.min(), y.max()],
    'r--',
    linewidth=2,
    label="Idealne dopasowanie (y = x)"
)

plt.xlabel("Wartość oczekiwana", fontsize=12)
plt.ylabel("Wartość prognozowana", fontsize=12)
plt.title("Porównanie wartości rzeczywistych i przewidywanych", fontsize=14)

# grid
plt.grid(True, linestyle='--', alpha=0.5)

# legend
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# imports
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
```

```python
# data
# input
X = np.random.rand(1000, 3) * [100000, 10, 20]

# output
y = np.random.rand(1000, 2) * [50000, 15]
```

```python
# neural network model definition
model = Sequential([
    Input(shape=(3,)),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(2)
])
```

```python
# model compilation
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
# network training
history = model.fit(
    X,
    y,
    epochs=100,
    batch_size=32,
    verbose=1
)
```

```python
# value prediction
y_pred = model.predict(X)
```

```python
# visualization
plt.figure(figsize=(7,7))

# scatter plot
plt.scatter(
    y[:,0],
    y_pred[:,0],
    alpha=0.6,
    s=40,
    edgecolors="black",
    label="Predykcje modelu"
)

# perfect fit line
plt.plot(
    [y[:,0].min(), y[:,0].max()],
    [y[:,0].min(), y[:,0].max()],
    'r--',
    linewidth=2,
    label="Idealne dopasowanie (y = x)"
)

plt.xlabel("Rzeczywisty zwrot z inwestycji", fontsize=12)
plt.ylabel("Prognozowany zwrot z inwestycji", fontsize=12)
plt.title("Porównanie wartości rzeczywistych i przewidywanych", fontsize=14)

# grid
plt.grid(True, linestyle="--", alpha=0.5)

# legend
plt.legend()
plt.tight_layout()
plt.show()
```
